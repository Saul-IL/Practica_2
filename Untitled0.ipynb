```python
n1=int(input("Ingrese su primer número "))
n2=int(input("Ingrese su segundo número "))
op=input("Ingrese en MAYUSCULAS y con sus acentos adecuados la operación que desea realizar ")

if op=="MULTIPLICACIÓN":
    print(n1*n2)

if op=="SUMA":
    print(n1+n2)

if op=="RESTA":
    print(n1-n2)

if op=="DIVISIÓN":
    print(n1/n2)
```
